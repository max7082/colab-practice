```python
import torch
```

```python
a=torch.tensor([2.0])
weights = [
    torch.tensor([i], dtype=torch.float32, requires_grad=True)
    for i in (1, 2, 3, 2)
]
w1, w2, w3, w4 = weights
b = w1 * a
c = w2 * a
d = w4 * b + w3 * c
loss = (15 - d) ** 2
loss.backward()
print("예측값:", d.item())
print("손실:", loss.item())
print("w1의 기울기:", w1.grad.item())
print("w2의 기울기:", w2.grad.item())
print("w3의 기울기:", w3.grad.item())
print("w4의 기울기:", w4.grad.item())
```
