# **Numpy**

```python
import numpy as np
import time
```

NumPy에서 제공되는 np.dot()함수와 직접 구현 함수의 속도 비교

```python
def vec_vec_mul_func(vec_x, vec_y):
  total=0
  for row in range(len(vec_x)):
    total+=vec_x[row]*vec_y[row]
  return total
```

```python
a=np.ones(10000)
b=np.ones(10000)
x = a.tolist()
y = b.tolist()
print(vec_vec_mul_func(x,y))
```

```python
def mat_vec_mul_func(mat_A, vec_x):
  result=[]
  for row in mat_A:
    total=0
    for col in range(len(row)):
      total+=row[col]*vec_x[col]
    result.append(total)
  return result
```

```python
a = np.ones((100, 100))
b = np.ones(100)
x = a.tolist()
y = b.tolist()
print(mat_vec_mul_func(x,y))
```

```python
def mat_mat_mul_func(mat_A, mat_B):
  result_mat=[]
  for row in mat_A:
    result_row=[]
    for col in range(len(mat_B[0])):
      total=0
      for i in range(len(row)):
        total+=row[i]*mat_B[i][col]
      result_row.append(total)
    result_mat.append(result_row)
  return result_mat
```

```python
a = np.ones((10, 10))
b = np.ones((10, 10))
x = a.tolist()
y = b.tolist()
print(mat_mat_mul_func(x,y))
```

```python
a=np.ones(10000)
b=np.ones(10000)
x = a.tolist()
y = b.tolist()
start = time.time()
np.dot(a, b)
end = time.time()
print("NumPy", end - start)
start = time.time()
vec_vec_mul_func(x, y)
end = time.time()
print("직접 구현:", end - start)
```

```python
a = np.ones((1000, 1000))
b = np.ones(1000)
x = a.tolist()
y = b.tolist()
start = time.time()
np.dot(a, b)
end = time.time()
print("NumPy:", end - start)

start = time.time()
mat_vec_mul_func(x, y)
end = time.time()
print("직접 구현:", end - start)
```
